```python
df['Last_Name']=df['Last_Name'].str.strip("123._/") # reg expression to remove unwanted characters from the last name column
```

```python
df['Phone_Number']
```

#### apply() => Apply is a Pandas function(Series Object) which will apply specified functionality to each value of a series 

```python
df['Phone_Number']=df['Phone_Number'].apply(lambda x : str(x))
```

```python
df['Phone_Number']
```

```python
df['Phone_Number']=df['Phone_Number'].str.replace('[^0-9]','',regex=True)
```

```python
df['Phone_Number']
```

```python
l1 = lambda x:x[0:3]+"-"+x[3:6]+"-"+x[6:10]
```

```python
df['Phone_Number']=df['Phone_Number'].apply(l1)
```

```python
df
```

```python
df['Address']
```

```python
# spliting address into 3 sep cols 
df[['Street_Addres','City','Pin_Code']] = df['Address'].str.split(',',expand=True)
```

```python
df
```

```python
df['Paying Customer']=df['Paying Customer'].str.replace('Yes','Y')
```

```python
df['Paying Customer']=df['Paying Customer'].str.replace('No','N')
```

```python
df['Paying Customer']
```

```python
df.columns
```

```python
df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace('Yes','Y')
df['Do_Not_Contact']=df['Do_Not_Contact'].str.replace('No','N')
```

```python
df['Do_Not_Contact']
```

```python
# remove columns 
df.drop(columns=['Not_Useful_Column','Address'],inplace=True)
```

```python
df
```

### Dealing with null values 

```python
df.replace('N/a','',inplace=True)
df.replace('NaN','',inplace=True)
```

```python
df.replace('--','',inplace=True)
```

```python
df.fillna('',inplace=True)
```

```python
for i in df.index:
    if df.loc[i,'Do_Not_Contact'] == "Y" or df.loc[i,'Phone_Number'] == "" :
        df.drop(i,inplace=True)
        
```

```python
df.reset_index(inplace=True,drop=True)
```

```python
df
```

```python
df.to_csv('Datasets/cleaned_ccl1.csv',index=False)
```
